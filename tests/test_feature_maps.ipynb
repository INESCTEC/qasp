#### 1) Angle embedding single-qubit features

```python

fm = AngleEmbedding(n_qubits=4, rotations="ZY")

qml.draw_mpl(fm)(inputs=np.random.random(4))
```

#### 1) Angle embedding arbitrary pauli string

```python

fm = AngleEmbedding(n_qubits=4, rotations=["YX 0 1", "Y 1", "ZX 0 2", "ZZZ 0 1 2"])


qml.draw_mpl(fm)(inputs=np.random.random(4))
```
